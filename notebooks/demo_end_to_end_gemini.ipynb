## Import system components
These imports reference your GitHub repository folder structure.

````python
from services.orchestrator.orchestrator import Orchestrator
from services.agents.market_research_agent import MarketResearchAgent
from services.agents.copy_agent import CopyAgent
from services.agents.webdev_agent import WebDevAgent

from rich.pretty import pprint
````

## Initialize Orchestrator & Register Gemini-Powered Agents

Each agent uses the Gemini API internally for its tasks.

````python
orch = Orchestrator()

orch.registry.register("MarketResearchAgent", MarketResearchAgent())
orch.registry.register("CopyAgent", CopyAgent())
orch.registry.register("WebDevAgent", WebDevAgent())

print("Agents registered:")
print(orch.registry.list())
````

## Run a Mission

This mission goes through:
1. Gemini-powered planning
2. Multi-agent execution
3. Judge evaluation
4. Auto-agent creation (if score < threshold)

---
🎯 **Mission:**
> Launch a full marketing campaign for a student-focused smartwatch

````python
mission = "Launch a full marketing campaign for a student-focused smartwatch"
result = orch.run_mission(mission)
pprint(result)
````

## Display Execution Trace

Showing each agent call and its output summary.

````python
for step in result['trace']:
    pprint(step)
    print("---")
````

## If Necessary: Run Again After Auto-Agent Creation

If the judge gave a low score, Aegis will generate a **new Gemini-powered agent**, usually `AnalyticsAgent`.

Running again lets you observe improvements.

````python
print("Re-running mission to test self-evolution...")
result2 = orch.run_mission(mission)
pprint(result2)
````
