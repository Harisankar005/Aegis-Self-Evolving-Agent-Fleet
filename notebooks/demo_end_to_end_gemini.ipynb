# Aegis – End-to-End Gemini Demo

This notebook demonstrates the complete mission flow of the Aegis multi-agent system using the Gemini API.

It performs:
- Mission input
- Planner → plan generation
- Execution of MarketResearchAgent, CopyAgent, WebDevAgent
- Session trace logging
- LLM-as-Judge scoring
- Automatic agent creation if score is low

## Imports and Initialization

```python
import os
from services.orchestrator.orchestrator import Orchestrator
from services.agents.market_research_agent import MarketResearchAgent
from services.agents.copy_agent import CopyAgent
from services.agents.webdev_agent import WebDevAgent
from rich.pretty import pprint

# API key must be set as environment variable inside Kaggle/Colab
assert os.getenv("GEMINI_API_KEY"), "GEMINI_API_KEY not set in environment variables."
```

## Initialize Orchestrator and Register Agents

```python
orch = Orchestrator()

orch.registry.register("MarketResearchAgent", MarketResearchAgent(), "Collects insights and competitor analysis.")
orch.registry.register("CopyAgent", CopyAgent(), "Generates landing page copy.")
orch.registry.register("WebDevAgent", WebDevAgent(), "Generates HTML landing page code.")

orch.registry.list()
```

## Define a Mission

```python
mission = "Launch a full marketing campaign for a student-focused smartwatch."
mission
```

## Run Mission End-to-End

```python
output = orch.run_mission(mission)
pprint(output)
```

## Display Session Trace
This shows the full sequence of agent events and outputs collected during execution.

```python
for event in output["trace"]:
    pprint(event)
    print("\n---\n")
```

## If Score Is Low, A New Agent Was Auto-Generated
The orchestrator automatically creates and registers a new agent when needed. You can verify:

```python
orch.registry.list()
```

## Re-run Mission After Auto-Generated Agent Is Added (Optional)

```python
output2 = orch.run_mission(mission)
pprint(output2)
```
